## Exact final theorem in common notation

Let

$$
f(z)=z^2+2,
$$

and let

$$
\mathcal B_\infty(f)=\{z\in\mathbb C : |f^{\circ n}(z)|\to\infty\}
$$

be the basin of infinity. Let $G_f$ be the
[Green function](https://en.wikipedia.org/wiki/Green%27s_function) of $f$, and
let $\Phi_\infty$ denote the normalized near-infinity
[Böttcher coordinate](https://en.wikipedia.org/wiki/B%C3%B6ttcher%27s_theorem).

The target `Quadratic.UnifiedGlobalBottcherTheoremFor (2 : ℂ)` amounts to the
existence of a map

$$
\Phi:\mathcal B_\infty(f)\to\mathbb C
$$

such that:

1. $\Phi$ is holomorphic on $\mathcal B_\infty(f)$;
2. $\Phi(f(z)) = \Phi(z)^2$ for all $z\in\mathcal B_\infty(f)$;
3. $|\Phi(z)| = e^{G_f(z)}$ on $\mathcal B_\infty(f)$;
4. $z\in\mathcal B_\infty(f)$ if and only if $|\Phi(z)|>1$;
5. $\Phi(z)/z\to 1$ as $z\to\infty$;
6. $\Phi$ agrees with $\Phi_\infty$ on a sufficiently large exterior region.


## Standard route from the literature

The theorem above is a standard result in holomorphic dynamics. The classical
proof proceeds in three steps.

**Step 1 (near-infinity Böttcher coordinate).** For any monic polynomial
$p$ of degree $d\ge 2$, the superattracting fixed point at infinity has a
unique normalized
[Böttcher coordinate](https://en.wikipedia.org/wiki/B%C3%B6ttcher%27s_theorem)
$\Phi_\infty$ satisfying $\Phi_\infty(p(z))=\Phi_\infty(z)^d$ and
$\Phi_\infty(z)/z\to 1$ on a neighborhood of infinity. This step is already
formalized in the repository as `genuineBottcherNearInfinityDataFor_logSeriesBottcherApprox`.

**Step 2 (extension to the full basin).** The key fact is that $\Phi_\infty$
extends to the full basin of infinity by iterative pullback:

$$
\Phi(z)=\Phi_\infty(p^{\circ N}(z))^{1/d^N}
$$

for any $N$ large enough that $p^{\circ N}(z)$ lies in the domain of
$\Phi_\infty$. The extension is well-defined because the filled Julia set
$K(p)$ is connected (equivalently, the critical point does not escape), which
forces the basin $\mathcal B_\infty(p)=\mathbb C\setminus K(p)$ to be simply
connected. Simple connectivity eliminates the $d^N$-th root ambiguity up to a
global rotation, and the normalization $\Phi(z)/z\to 1$ fixes the rotation.

The primary references are:

- Douady and Hubbard, *Étude dynamique des polynômes complexes* Part I, §7;
- Milnor, *Dynamics in One Complex Variable*, §9 (Theorem 9.1 and
  Corollary 9.4);
- Carleson and Gamelin, *Complex Dynamics*, Chapter III.

**Step 3 (properties of the extension).** The extended $\Phi$ is a conformal
isomorphism $\mathcal B_\infty(p)\xrightarrow{\sim}\{w:|w|>1\}$, and
$\log|\Phi(z)|$ equals the
[Green function](https://en.wikipedia.org/wiki/Green%27s_function) $G_p(z)$.
The semiconjugacy $\Phi(p(z))=\Phi(z)^d$ and the basin characterization
$z\in\mathcal B_\infty(p)\iff|\Phi(z)|>1$ are immediate consequences.

For $f(z)=z^2+2$ the filled Julia set is the real interval $[-2,2]$ (connected),
so all hypotheses are satisfied and the theorem applies directly.
