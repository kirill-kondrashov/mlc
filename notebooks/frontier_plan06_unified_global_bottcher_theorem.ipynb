Let

```math
f(z)=z^2+2
```

and let

```math
\mathcal B_\infty(f)=\{z\in\mathbf C: |f^{\circ N}(z)|\to\infty\text{ as }N\to\infty\}
```

be the basin of infinity. Let $\Phi_\infty$ be the already formalized near-infinity Böttcher coordinate on an exterior region.

**Problem.** Construct a holomorphic coordinate

```math
\Phi:\mathcal B_\infty(f)\to\{w\in\mathbf C: |w|>1\}
```

such that:

1. $\Phi$ agrees with $\Phi_\infty$ near infinity;
2. $\Phi(f(z))=\Phi(z)^2$ on $\mathcal B_\infty(f)$;
3. $z\in\mathcal B_\infty(f)$ if and only if $1<|\Phi(z)|$;
4. $\Phi$ satisfies the Green-function modulus identity and normalization at infinity used by the Lean structure `LogSeriesBasinExtensionDataFor`;
5. the same map supplies the inverse package required by `GenuineBottcherInversePackageFor`.

In Lean terms, this is the construction of `Quadratic.UnifiedGlobalBottcherTheoremFor (2 : ℂ)`.
