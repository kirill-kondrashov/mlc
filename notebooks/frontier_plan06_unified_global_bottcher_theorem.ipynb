**Adjusted frontier theorem.** For every such loop $\gamma$, there exist:

$$
c_{jk}=b_k-b_j\in 2\pi i\mathbf Z,

## Exact rigorous proof status

A full rigorous proof of the **Adjusted frontier theorem** is **not currently available** in the repository. The obstruction is not the Cech-closing step or the overlap-comparison bookkeeping; those are already formalized. The missing part is the genuinely dynamical construction of the anchored global logarithm on a zero-free neighborhood of the actual high-escaping image.

What can be proved rigorously right now is the following exact reduction.

**Proposition (rigorous reduction to the anchored global-log datum).** Fix an admissible high-escaping chart-chain loop $\gamma$. Assume that for some sufficiently large level $K$ there exist:

1. a connected open set
   $$
   \Omega\subset\mathbf C^*,\qquad A_K([0,1])\subset\Omega;
   $$
2. a holomorphic logarithm
   $$
   L:\Omega\to\mathbf C,\qquad e^{L(w)}=w\text{ on }\Omega;
   $$
3. a nonempty anchor set $B\subset\Omega$ on which $L$ agrees with the normalized near-infinity logarithm germ used to define the actual branches;
4. an actual finite chart cover $(U_j,V_j,b_j)_{0\le j\le m}$ of $A_K([0,1])$, refined so that every $U_j\subset\Omega$ and every overlap neighborhood lies in the corresponding chart intersection.

Then the conclusion of the Adjusted frontier theorem holds for $\gamma$: after refinement one has

$$
b_j=L|_{U_j}\qquad\text{on }U_j,
$$

the transition cocycle

$$
c_{jk}=b_k-b_j
$$

is zero, and hence

$$
b_m=b_0\qquad\text{on }V_{\mathrm{cl}}.
$$

**Proof.** Since $A_K([0,1])$ is compact and contained in the open set $\Omega$, the actual cover may be refined so that each $U_j$ is connected, zero-free, contained in $\Omega$, and every overlap neighborhood $V_j$ and $V_{\mathrm{cl}}$ remains inside the relevant chart intersection. On each $U_j$, both $b_j$ and $L|_{U_j}$ are holomorphic logarithms of the identity map, hence

$$
e^{b_j(w)}=w=e^{L(w)}\qquad (w\in U_j).
$$

By the anchoring hypothesis, the actual branch $b_j$ and the restriction $L|_{U_j}$ agree on the nonempty set where the continuation path issuing from the normalized near-infinity germ first enters $U_j$. Since $U_j$ is connected and both functions are holomorphic, the identity theorem gives

$$
b_j=L|_{U_j}\qquad\text{on }U_j.
$$

Therefore on every overlap one has

$$
c_{jk}=b_k-b_j=L|_{U_k}-L|_{U_j}=0.
$$

So the transition cocycle is the zero cocycle, hence in particular a Cech coboundary. Applying the already checked closing theorem `Quadratic.logBranch_eqOn_closing_of_coboundary` with the choice $a_j=0$ for all $j$ yields

$$
b_m=b_0\qquad\text{on }V_{\mathrm{cl}}.
$$

This proves the adjusted theorem **conditional on the anchored global-log datum**. $\square$

So the exact rigorous status is now sharp:

- the reduction from one anchored global logarithm to cocycle vanishing and closing equality is rigorous;
- the missing unproved step is the dynamical theorem producing that anchored global logarithm from the actual high-escaping geometry.


## Numerical probe of the missing anchored-global-log step

The missing theorem is the existence, for every admissible high-escaping chart-chain loop, of one connected zero-free neighborhood $\Omega$ carrying one anchored holomorphic logarithm for the whole actual high-level image. This can be tested numerically on concrete loop families by examining the loop

$$
A_K(t)=\Phi_\infty\bigl(f^{\circ K}(\gamma(t))\bigr).
$$

A positive numerical example should exhibit a visible zero-free region containing $A_K([0,1])$ on which the principal logarithm already works. A counterexample candidate should exhibit nonzero winding of $A_K$ about $0$, since then no connected open set $\Omega\subset\mathbf C^*$ containing the image can carry a holomorphic logarithm.

### Positive numerical example

Take

$$
\gamma_{+}(t)=0.2e^{2\pi i t},\qquad 0\le t\le 1,
$$

which is a small basin loop around the escaping point $0$. At level $K=2$ one finds numerically:

1. $\min_t |f^{\circ 2}(\gamma_{+}(t))|\approx 5.841600>4$, so the whole second iterate already lies in the checked exterior region;
2. $\min_t |A_2(t)|\approx 6.012651$;
3. $\min_t \Re A_2(t)\approx 6.012651$ and $\max_t |\Im A_2(t)|\approx 0.155607$;
4. $\operatorname{wind}(A_2,0)\approx 0$;
5. the unwrapped principal logarithm closes up with error about $1.23\times 10^{-17}$.

So in this example the whole image $A_2([0,1])$ sits numerically inside the right half-plane $\{\Re w>5\}$, and the principal branch of $\log$ on that half-plane gives a concrete anchored logarithm candidate. This is strong positive evidence that the missing theorem is at least plausible on some genuine nontrivial basin loops that are not already wholly contained in the outside region at level $0$.

### Attempted numerical falsification

Now return to the historical precritical family

Hence no connected open neighborhood $\Omega\subset\mathbf C^*$ containing this numerical image can carry a single-valued holomorphic logarithm. So this remains a genuine **counterexample candidate to the missing anchored-global-log theorem**.

However, it is **not yet a falsification of the Adjusted frontier theorem itself**, because the theorem only quantifies over the smaller class of admissible high-escaping chart-chain loops actually used by the PLAN 08/09 descent argument. ## Why the older all-basin-loop theorem was too strong

The previous frontier formulation asked for trivial $2^{-N}$-root monodromy on **every** basin loop. The new numerical experiments strongly suggest that this unrestricted statement is false: loops surrounding precritical points such as $\pm i\sqrt{2}$ can carry nontrivial monodromy multipliers.

The key correction is that MLC does not need control of all basin loops. The PLAN 08/09 descent argument only uses the loops that arise from the **actual finite overlap cycle** on a sufficiently high escaping image, where continuation is performed chart-by-chart and then closed along one finite covering cycle.

So the relevant class is:

1. basin loops $\gamma$ for which some sufficiently high level $K$ has
   $$
   A_K([0,1])\subset\mathbf C^*;
   $$
2. whose high-level image is handled by the actual finite chart cover from PLAN 08/09;
3. and for which one only needs to compare branches on the consecutive overlaps and the closing overlap of that finite cover.

This is the candidate class of **admissible high-escaping chart-chain loops**. The adjusted theorem from the previous section says exactly what should be proved for this class: not that every basin loop has trivial pullback monodromy, but that every actual high-escaping chart chain used by the descent argument is controlled by one anchored logarithm on one zero-free neighborhood.

For reference, the older unrestricted statement was the following.

Write

$$
\Omega_R=\{z\in\mathbf C:|z|>R\},\qquad R=4,
$$

and let

$$
\Phi_\infty:\Omega_R\to\{w\in\mathbf C:|w|>1\}
$$

be the checked near-infinity Bottcher coordinate satisfying

$$
\Phi_\infty(f(z))=\Phi_\infty(z)^2
\qquad (z\in\Omega_R).
$$

Let a **basin loop** be a continuous map

$$
\gamma:[0,1]\to\mathcal B_\infty(f),\qquad \gamma(0)=\gamma(1).
$$

Fix an integer $N\geq 0$ such that the whole image of the $N$-th iterate of the loop lies in the exterior region:

$$
f^{\circ N}(\gamma([0,1]))\subset \Omega_R.
$$

Then define the exterior-value loop

$$
A_{\gamma,N}(t)=\Phi_\infty\bigl(f^{\circ N}(\gamma(t))\bigr),\qquad 0\le t\le 1.
$$

Since $|A_{\gamma,N}(t)|>1$, this is a loop in $\mathbf C^\times$. A continuous **pullback branch at level $N$** along $\gamma$ means a continuous map

$$
B_{\gamma,N}:[0,1]\to\mathbf C^\times
$$

such that

$$
B_{\gamma,N}(t)^{2^N}=A_{\gamma,N}(t)
\qquad (0\le t\le 1).
$$

The pullback branch has **trivial monodromy** if every such continuation closes up:

$$
B_{\gamma,N}(1)=B_{\gamma,N}(0).
$$

Equivalently, if $A_{\gamma,N}$ is piecewise $C^1$, then trivial monodromy is equivalent to the divisibility condition

$$
\operatorname{wind}(A_{\gamma,N},0)\in 2^N\mathbf Z,
$$

where

$$
\operatorname{wind}(A_{\gamma,N},0)=\frac{1}{2\pi i}\int_0^1 \frac{A_{\gamma,N}'(t)}{A_{\gamma,N}(t)}\,dt.
$$

The older theorem asked that for every basin loop $\gamma$ and every integer $N\geq 0$ with

$$
f^{\circ N}(\gamma([0,1]))\subset\Omega_R,
$$

the loop $A_{\gamma,N}$ should have trivial $2^{-N}$-root monodromy. Equivalently,

$$
\operatorname{wind}\!\left(\Phi_\infty\circ f^{\circ N}\circ\gamma,0\right)\in 2^N\mathbf Z.
$$

The experiments below should therefore be read historically: they are evidence against this older unrestricted theorem, and the reason the live frontier has been narrowed to the actual high-escaping chart-chain theorem stated above.


## Numerical evidence against the older global-loop formulation

The older unrestricted theorem would imply a single-valued pullback branch on every basin loop.

To test this directly, we look at the loop

$$
\gamma(t)=i\sqrt{2}+0.8e^{2\pi i t},\qquad 0\le t\le 1.
$$

This loop surrounds the simple preimage $i\sqrt{2}$ of $0$ and, numerically, its third iterate already lies entirely in the checked outside region $|z|>4$.

If the older unrestricted theorem were true, then along this loop the value $\Phi(\gamma(t))$ would have to be a continuous $2^3$-rd root of

$$
A_3(t)=\Phi_\infty\bigl(f^{\circ 3}(\gamma(t))\bigr).
$$

So if $A_3$ winds around $0$ by a number not divisible by $2^3=8$, analytic continuation of an $8$-th root returns multiplied by a nontrivial root of unity, and the required single-valued branch cannot exist on this loop.


```python
import cmath
import math
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 120

c = 2.0 + 0.0j
R = 4.0

def f(z):
    return z * z + c

def iterate(z, n):
    for _ in range(n):
        z = f(z)
    return z

def phi_inf_approx(z, terms=60):
    s = 0.0 + 0.0j
    w = z
    for n in range(terms):
        s += (0.5 ** (n + 1)) * cmath.log(1 + c / (w * w))
        w = f(w)
        if abs(w) > 1e120:
            break
    return z * cmath.exp(s)

def winding_number(path):
    phases = np.unwrap(np.angle(path))
    return (phases[-1] - phases[0]) / (2 * np.pi)

def continued_root(A_path, degree):
    phases = np.unwrap(np.angle(A_path)) / degree
    radii = np.abs(A_path) ** (1 / degree)
    return radii * np.exp(1j * phases)

samples = 2400
t = np.linspace(0.0, 1.0, samples)
theta = 2 * np.pi * t

center = 1j * math.sqrt(2.0)
radius = 0.8
N = 3
degree = 2 ** N

gamma = center + radius * np.exp(1j * theta)
iter3 = np.array([iterate(z, N) for z in gamma])
A3 = np.array([phi_inf_approx(z) for z in iter3])
root_path = continued_root(A3, degree)

wind_iter3 = winding_number(iter3)
wind_A3 = winding_number(A3)
multiplier = root_path[-1] / root_path[0]
expected_multiplier = cmath.exp(2j * math.pi * round(wind_A3) / degree)
min_abs_iter3 = float(np.min(np.abs(iter3)))

fig, axes = plt.subplots(1, 3, figsize=(16.5, 5.0))

ax = axes[0]
ax.plot(gamma.real, gamma.imag, color='tab:blue', lw=2)
ax.scatter([center.real], [center.imag], color='crimson', s=70, label=r'$i\sqrt{2}$')
ax.set_aspect('equal', adjustable='box')
ax.set_title(r'Basin loop $\gamma$ around $i\sqrt{2}$')
ax.set_xlabel(r'$\Re z$')
ax.set_ylabel(r'$\Im z$')
ax.legend(loc='upper right')

ax = axes[1]
circle = R * np.exp(1j * theta)
ax.plot(iter3.real, iter3.imag, color='tab:green', lw=2, label=r'$f^{\circ 3}(\gamma)$')
ax.plot(circle.real, circle.imag, 'k--', lw=1.0, label=r'$|z|=4$')
ax.set_aspect('equal', adjustable='box')
ax.set_title(r'Third iterate already lies in the outside region')
ax.set_xlabel(r'$\Re z$')
ax.set_ylabel(r'$\Im z$')
ax.legend(loc='upper right')

ax = axes[2]
ax.plot(root_path.real, root_path.imag, color='tab:purple', lw=2, label='continued 8th root')
ax.scatter([root_path[0].real], [root_path[0].imag], color='black', s=65, label='start')
ax.scatter([root_path[-1].real], [root_path[-1].imag], color='crimson', marker='x', s=85, label='end')
ax.set_aspect('equal', adjustable='box')
ax.set_title('Analytic continuation of the required root')
ax.set_xlabel(r'$\Re w$')
ax.set_ylabel(r'$\Im w$')
ax.legend(loc='upper right')

print('Unified-theorem numerical counterexample candidate')
print('loop center =', center)
print('loop radius =', radius)
print('escape level N =', N, 'root degree =', degree)
print('min |f^N(gamma)| =', min_abs_iter3)
print('winding of f^N(gamma) about 0 =', wind_iter3)
print('winding of A_N = Phi_inf(f^N(gamma)) about 0 =', wind_A3)
print('continued-root multiplier =', multiplier)
print('expected multiplier from winding =', expected_multiplier)
print('is the required root single-valued on the loop?', abs(multiplier - 1) < 1e-8)
plt.show()
```

On the left, the loop $\gamma$ sits around the escaping preimage $i\sqrt{2}$, so every point on the loop is numerically in the basin of infinity. In the middle panel, the third iterate $f^{\circ 3}(\gamma)$ lies strictly outside the dashed circle $|z|=4$, so this is already a valid escape level for the checked near-infinity coordinate.

The key number is the winding of
$$
A_3(t)=\Phi_\infty\bigl(f^{\circ 3}(\gamma(t))\bigr).
$$
Numerically it is $4$. Since the required pullback root has degree $2^3=8$, analytic continuation of an $8$-th root around the loop returns with multiplier
$$
e^{2\pi i \cdot 4 / 8} = -1,
$$
not with multiplier $1$. This is what the right-hand panel shows: the continued root starts at one point and ends at its negative.

So this is a direct **numerical counterexample candidate** to the unified frontier theorem. If these windings reflect the exact basin monodromy, then no single-valued global map $\Phi$ with the stated properties can exist on the actual basin, because Theorem A already fails on this loop and hence the stronger unified theorem fails as well.


## Organized experiment sweep

To see whether the obstruction above is merely a one-off sample or a stable monodromy phenomenon, we run three focused numerical experiments:

1. **Resolution stability** for the original loop $\gamma(t)=i\sqrt{2}+0.8e^{2\pi i t}$ at escape level $N=3$.
2. **Radius regime sweep** for loops around both simple preimages $\pm i\sqrt{2}$, recording the first level $N$ at which the whole iterated loop lies outside $|z|>4$.
3. **Persistence across escape levels** for the robust loop of radius $0.8$, checking whether the nontrivial multiplier disappears when one pulls back from deeper exterior levels.

The experiment is successful only in the numerical sense if the same nontrivial root multiplier keeps appearing across these variations. It is successful in the rigorous mathematical sense only if the numerics can later be upgraded to exact winding statements on genuine basin loops.


```python
import cmath
import math
import numpy as np

c = 2.0 + 0.0j
R = 4.0
phi_terms = 80

def f(z):
    return z * z + c

def iterate(z, n):
    for _ in range(n):
        z = f(z)
    return z

def phi_inf_approx(z, terms=phi_terms):
    s = 0.0 + 0.0j
    w = z
    for n in range(terms):
        s += (0.5 ** (n + 1)) * cmath.log(1 + c / (w * w))
        w = f(w)
        if abs(w) > 1e120:
            break
    return z * cmath.exp(s)

def winding_number(path):
    phases = np.unwrap(np.angle(path))
    return (phases[-1] - phases[0]) / (2 * np.pi)

def continued_root(A_path, degree):
    phases = np.unwrap(np.angle(A_path)) / degree
    radii = np.abs(A_path) ** (1 / degree)
    return radii * np.exp(1j * phases)

def loop_data(center, radius, N, samples=2400):
    t = np.linspace(0.0, 1.0, samples)
    gamma = center + radius * np.exp(2j * np.pi * t)
    iterN = np.array([iterate(z, N) for z in gamma])
    A = np.array([phi_inf_approx(z) for z in iterN])
    degree = 2 ** N
    root_path = continued_root(A, degree)
    multiplier = root_path[-1] / root_path[0]
    return {
        'min_abs': float(np.min(np.abs(iterN))),
        'wind_A': float(winding_number(A)),
        'degree': degree,
        'multiplier': multiplier,
        'single': abs(multiplier - 1) < 1e-6,
    }

def first_outside(center, radius, maxN=6, samples=2400):
    for N in range(1, maxN + 1):
        data = loop_data(center, radius, N, samples=samples)
        if data['min_abs'] > R:
            return N, data
    return None, None

def fmt_complex(z):
    return f"{z.real:.6f}{z.imag:+.6f}i"

center = 1j * math.sqrt(2.0)
print('Organized monodromy experiment sweep')
print('')
print('1. Resolution stability for center = i*sqrt(2), radius = 0.8, N = 3')
print('   samples   wind(A_N)      multiplier                 single-valued?')
for samples in [600, 1200, 2400, 4800]:
    data = loop_data(center, 0.8, 3, samples=samples)
    print(f"   {samples:6d}   {data['wind_A']:10.6f}   {fmt_complex(data['multiplier']):>24}   {data['single']}")

print('')
print('2. Radius regime sweep (first escape level with full iterate in |z| > 4)')
for center in [1j * math.sqrt(2.0), -1j * math.sqrt(2.0)]:
    print(f"   center = {center}")
    print('   radius   first N   degree   min|f^N(gamma)|   wind(A_N)    multiplier                 single-valued?')
    for radius in [0.2, 0.4, 0.5, 0.8, 1.0]:
        N, data = first_outside(center, radius, maxN=6, samples=2400)
        if N is None:
            print(f"   {radius:6.2f}   none      --       --               --          --                        --")
        else:
            print(
                f"   {radius:6.2f}   {N:7d}   {data['degree']:6d}   {data['min_abs']:15.6f}   "
                f"{data['wind_A']:10.6f}   {fmt_complex(data['multiplier']):>24}   {data['single']}"
            )

print('')
print('3. Persistence across deeper escape levels for center = i*sqrt(2), radius = 0.8')
print('   N   degree   min|f^N(gamma)|   wind(A_N)    multiplier                 single-valued?')
for N in [3, 4, 5]:
    data = loop_data(center, 0.8, N, samples=2400)
    print(
        f"   {N:1d}   {data['degree']:6d}   {data['min_abs']:15.6f}   {data['wind_A']:10.6f}   "
        f"{fmt_complex(data['multiplier']):>24}   {data['single']}"
    )
```

### Conclusion from the organized experiments

The experiment is **successful as numerical evidence for a monodromy obstruction**, but **not successful as a rigorous falsification** of the older unrestricted theorem.

The evidence is stronger than the original one-loop plot alone:

1. the original loop at radius $0.8$ is resolution-stable, with winding $4$ and continued-root multiplier numerically equal to $-1$ from $600$ up to $4800$ samples;
2. the same nontrivial behavior appears for loops around both simple preimages $\pm i\sqrt{2}$;
3. once one enters the nontrivial regime, the obstruction persists at deeper escape levels rather than disappearing;
4. the phenomenon is not all-or-nothing: small loops such as radii $0.2$ and $0.4$ have trivial monodromy, while larger loops numerically produce multipliers such as $-1$ and even $e^{\pi i/4}$.

So the current numerics strongly suggest that the historical theorem from the previous section is false in its unrestricted form: not every basin loop appears to admit trivial $2^{-N}$-root monodromy once the iterated loop reaches the exterior region.

However, this still falls short of a proof. What is missing is an exact argument that the tested loops are genuine basin loops and that the measured windings are exact topological winding numbers rather than stable numerical approximations. Until that upgrade is made, the correct conclusion is:

- **successful numerical obstruction candidate against the older global-loop theorem:** yes;
- **rigorous falsification of that older theorem:** not yet.

These experiments do **not** directly attack the adjusted frontier theorem of this notebook, because that theorem only quantifies over the actual high-escaping chart-chain class and only asks for one anchored logarithm on the actual finite cover used downstream.
