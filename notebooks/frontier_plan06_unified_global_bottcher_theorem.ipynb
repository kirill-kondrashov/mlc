## ⚠️ CRITICAL FINDING: the target is mathematically unsatisfiable

**The statement `UnifiedGlobalBottcherTheoremFor (2 : ℂ)` is mathematically false.**

### Why $c=2$ makes the target impossible

For $f(z)=z^2+2$ the critical orbit $0\to 2\to 6\to 38\to\cdots$ escapes,
so $c=2\notin\mathcal M$. The filled Julia set $K(f)$ is a
[Cantor set](https://en.wikipedia.org/wiki/Cantor_set) (totally disconnected).

By Alexander duality in the plane, the complement of a compact totally
disconnected set has **nontrivial fundamental group**:

$$
\pi_1(\mathcal B_\infty(f))\ne 0.
$$

The classical Böttcher extension proceeds by choosing branches of
$\Phi(z) = \Phi_\infty(f^{\circ N}(z))^{1/2^N}$. On a non-simply-connected
domain the $2^N$-th root introduces nontrivial monodromy: a loop around a
component of $J(f)$ multiplies the branch by $e^{2\pi i/2^k}$.

**Consequence:** No single-valued holomorphic function
$\Phi:\mathcal B_\infty(f)\to\mathbb C$ can simultaneously satisfy the
semiconjugacy $\Phi(f(z))=\Phi(z)^2$ and extend the near-infinity Böttcher
coordinate to the full basin. The fields of `LogSeriesBasinExtensionDataFor`
(in particular `holo_on_basin : DifferentiableOn ℂ φ (basin_of_infinity c)`)
are unsatisfiable for $c=2$.

### Classical fact

For $c\notin\mathcal M$ the Böttcher coordinate extends holomorphically only to
$\{G_c(z)>G_c(c)\}$ (above the critical value equipotential), which IS simply
connected. Below that level, single-valued extension is impossible due to the
branching at the critical value.

References:

- Milnor, *Dynamics in One Complex Variable*, §9, Remark after Theorem 9.1;
- Douady and Hubbard, *Étude dynamique* I, §7.3.

## Standard route from the literature (for $c\in\mathcal M$)

The classical Böttcher extension theorem applies when the filled Julia set is
connected ($c\in\mathcal M$). For reference, the proof proceeds in three steps.

**Step 1 (near-infinity Böttcher coordinate).** For any monic polynomial
$p$ of degree $d\ge 2$, the superattracting fixed point at infinity has a
unique normalized
[Böttcher coordinate](https://en.wikipedia.org/wiki/B%C3%B6ttcher%27s_theorem)
$\Phi_\infty$ satisfying $\Phi_\infty(p(z))=\Phi_\infty(z)^d$ and
$\Phi_\infty(z)/z\to 1$ on a neighborhood of infinity. This step is already
formalized in the repository as `genuineBottcherNearInfinityDataFor_logSeriesBottcherApprox`.

**Step 2 (extension to the full basin).** The key fact is that $\Phi_\infty$
extends to the full basin of infinity by iterative pullback:

$$
\Phi(z)=\Phi_\infty(p^{\circ N}(z))^{1/d^N}
$$

for any $N$ large enough that $p^{\circ N}(z)$ lies in the domain of
$\Phi_\infty$. The extension is well-defined because the filled Julia set
$K(p)$ is connected (equivalently, the critical point does not escape), which
forces the basin $\mathcal B_\infty(p)=\mathbb C\setminus K(p)$ to be simply
connected. Simple connectivity eliminates the $d^N$-th root ambiguity up to a
global rotation, and the normalization $\Phi(z)/z\to 1$ fixes the rotation.

References: Douady–Hubbard I §7; Milnor §9 (Theorem 9.1); Carleson–Gamelin Ch. III.

**Step 3 (properties of the extension).** The extended $\Phi$ is a conformal
isomorphism $\mathcal B_\infty(p)\xrightarrow{\sim}\{w:|w|>1\}$, and
$\log|\Phi(z)|$ equals the Green function $G_p(z)$.
The semiconjugacy $\Phi(p(z))=\Phi(z)^d$ and the basin characterization
$z\in\mathcal B_\infty(p)\iff|\Phi(z)|>1$ are immediate consequences.

**This argument does NOT apply to $c=2$ because the critical point escapes
and $K(f_2)$ is disconnected.**
