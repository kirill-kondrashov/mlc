# PLAN 08: geometry of the missing monodromy theorem

This notebook visualizes the theorem still missing behind PLAN 08.

Lean now has theorem surfaces for:

```lean
BasinLoop
LocalPullbackRootBranchData
AnalyticContinuationAlongBasinLoop
BasinLoopPullbackRootMonodromyData
```

but not their actual analytic construction.

The missing theorem says:

> local pullback root branches continue analytically along every basin loop, the result differs by a root-of-unity multiplier, and for the normalized Böttcher branch that multiplier is always $1$.

The visualization below separates the geometry into three layers:

1. a loop in the basin;
2. the image loop $A_N(z)=\Phi_\infty(f^N(z))$ in the punctured complex plane;
3. the continued $2^N$-root branch above that image loop.

````python
import cmath
import math
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.dpi'] = 120

c = 2.0 + 0.0j

def f(z):
    return z*z + c

def iterate(z, n):
    for _ in range(n):
        z = f(z)
    return z

def phi_inf_approx(z, terms=50):
    s = 0.0 + 0.0j
    w = z
    for n in range(terms):
        s += (0.5 ** (n + 1)) * cmath.log(1 + c / (w*w))
        w = f(w)
        if abs(w) > 1e80:
            break
    return z * cmath.exp(s)

def continued_root(A_path, degree):
    phases = np.unwrap(np.angle(A_path)) / degree
    radii = np.abs(A_path) ** (1 / degree)
    return radii * np.exp(1j * phases)

def winding_number(path):
    return (np.unwrap(np.angle(path))[-1] - np.unwrap(np.angle(path))[0]) / (2*np.pi)
````

## A model basin loop and its $A_N$ image

For a genuine proof, the loop must live in the basin $U_\infty$. Here we use a controlled model loop to show the geometry of what Lean must formalize.

Given a loop $\gamma(t)$, define

$$
A_N(t)=\Phi_\infty(f^N(\gamma(t))).
$$

At level $N$, a local branch is a function $W_N(t)$ satisfying

$$
W_N(t)^{2^N}=A_N(t).
$$

The monodromy is the multiplier

$$
W_N(1)=\rho_N(\gamma)W_N(0),
\qquad \rho_N(\gamma)\in\mu_{2^N}.
$$

The missing theorem is $\rho_N(\gamma)=1$ for the normalized branch.

````python
# Model loop in the z-plane. We choose a loop far enough out so the near-infinity
# approximation is meaningful and the picture is stable.
t = np.linspace(0, 2*np.pi, 800)
z_loop = 5.0 + 0.8 * np.exp(1j * t)
N = 2
A_loop = np.array([phi_inf_approx(iterate(z, N), terms=60) for z in z_loop])
degree = 2**N
root_loop = continued_root(A_loop, degree)

rho = root_loop[-1] / root_loop[0]
wind_A = winding_number(A_loop)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

ax = axes[0]
ax.plot(z_loop.real, z_loop.imag, color='tab:blue')
ax.scatter([z_loop[0].real], [z_loop[0].imag], color='green', s=70, label='start=end')
ax.set_aspect('equal', adjustable='box')
ax.set_title('1. loop gamma in the basin region')
ax.set_xlabel('Re z')
ax.set_ylabel('Im z')
ax.legend()

ax = axes[1]
ax.plot(A_loop.real, A_loop.imag, color='tab:red')
ax.scatter([0], [0], color='black', s=25, label='0')
ax.scatter([A_loop[0].real], [A_loop[0].imag], color='green', s=70, label='start=end')
ax.set_aspect('equal', adjustable='box')
ax.set_title('2. image loop A_N(t) in C*')
ax.set_xlabel('Re A_N')
ax.set_ylabel('Im A_N')
ax.legend()

ax = axes[2]
ax.plot(root_loop.real, root_loop.imag, color='limegreen')
ax.scatter([root_loop[0].real], [root_loop[0].imag], color='green', s=70, label='start')
ax.scatter([root_loop[-1].real], [root_loop[-1].imag], color='red', marker='x', s=90, label='end')
ax.set_aspect('equal', adjustable='box')
ax.set_title(f'3. continued {degree}-th root branch')
ax.set_xlabel('Re W_N')
ax.set_ylabel('Im W_N')
ax.legend()

fig.tight_layout()
plt.show()

print(f'level N = {N}, root degree = {degree}')
print(f'winding number of A_N loop around 0 ≈ {wind_A:.6g}')
print(f'monodromy multiplier rho = W_N(end)/W_N(start) ≈ {rho:.8g}')
print(f'is rho approximately 1? {abs(rho - 1) < 1e-6}')
````

## What bad monodromy would look like

The previous loop is not meant as a counterexample; it is a geometric diagnostic. To see the obstruction clearly, compare with an artificial loop $A(t)$ that winds around $0$.

If $A(t)$ winds once and the root degree is $2^N$, the continued root returns multiplied by

$$
\exp(2\pi i/2^N).
$$

That is a nontrivial element of $\mu_{2^N}$, so the branch is not single-valued downstairs.

````python
degrees = [2, 4, 8]
t = np.linspace(0, 2*np.pi, 800)
A_bad = np.exp(1j * t)

fig, axes = plt.subplots(1, len(degrees), figsize=(15, 4.5))
for ax, degree in zip(axes, degrees):
    root = continued_root(A_bad, degree)
    rho = root[-1] / root[0]
    ax.plot(root.real, root.imag, color='limegreen')
    ax.scatter([root[0].real], [root[0].imag], color='green', s=70, label='start')
    ax.scatter([root[-1].real], [root[-1].imag], color='red', marker='x', s=90, label='end')
    ax.set_aspect('equal', adjustable='box')
    ax.set_title(f'degree {degree}, rho~{rho:.3g}')
    ax.set_xlabel('Re root')
    ax.set_ylabel('Im root')
    ax.legend(fontsize=8)
fig.suptitle('Artificial winding loop: nontrivial root monodromy')
fig.tight_layout()
plt.show()
````

## Special case proof: image loop lies in a simply connected disk avoiding zero

A clean special case of the missing theorem is the following.

Suppose the loop

$$
A_N(t)=\Phi_\infty(f^N(\gamma(t)))
$$

is contained in an open disk $D(a,r)$ such that

$$
0\notin D(a,r).
$$

Then $D(a,r)$ is simply connected and avoids zero, so there is a single holomorphic logarithm branch on that disk. Define

$$
W_N(t)=\exp\left(\frac{1}{2^N}Log_D(A_N(t))\right).
$$

Because the same logarithm branch is used for the whole loop, $W_N(t)$ returns to its starting value. Hence the monodromy is trivial in this special case.

This is not the full basin theorem, because for a general basin loop the image $A_N(\gamma)$ may not be contained in one such disk. But it gives a rigorous local pattern:

1. find a simply connected zero-free neighborhood of the image;
2. choose a logarithm branch there;
3. define the root using that logarithm;
4. monodromy is automatically trivial.

````python
# Special case: a loop inside a disk centered at a=2, radius r=0.5.
# This disk avoids 0, so a single log branch exists.
a = 2.0 + 0.0j
r = 0.5
degree = 8
t = np.linspace(0, 2*np.pi, 600)
A_good = a + r * np.exp(1j * t)

# Use the logarithm branch Log_D induced by the principal log; since the disk
# stays in the right half-plane and avoids 0, this is continuous on the disk.
W_good = np.exp(np.log(A_good) / degree)
rho_good = W_good[-1] / W_good[0]

# Compare with a bad loop enclosing 0.
A_bad = np.exp(1j * t)
W_bad = continued_root(A_bad, degree)
rho_bad = W_bad[-1] / W_bad[0]

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

ax = axes[0, 0]
ax.plot(A_good.real, A_good.imag, color='tab:blue', linewidth=2)
disk = plt.Circle((a.real, a.imag), r, fill=False, color='tab:blue', linestyle='--', label='zero-free disk')
ax.add_patch(disk)
ax.scatter([0], [0], color='black', s=30, label='0')
ax.set_aspect('equal', adjustable='box')
ax.set_title('Good case: A-loop inside disk avoiding 0')
ax.set_xlabel('Re A')
ax.set_ylabel('Im A')
ax.legend()

ax = axes[0, 1]
ax.plot(W_good.real, W_good.imag, color='limegreen', linewidth=2)
ax.scatter([W_good[0].real], [W_good[0].imag], color='green', s=70, label='start')
ax.scatter([W_good[-1].real], [W_good[-1].imag], color='red', marker='x', s=90, label='end')
ax.set_aspect('equal', adjustable='box')
ax.set_title(f'Root using one log branch, rho~{rho_good:.3g}')
ax.set_xlabel('Re root')
ax.set_ylabel('Im root')
ax.legend()

ax = axes[1, 0]
ax.plot(A_bad.real, A_bad.imag, color='tab:red', linewidth=2)
ax.scatter([0], [0], color='black', s=30, label='0')
ax.set_aspect('equal', adjustable='box')
ax.set_title('Bad case: A-loop encloses 0')
ax.set_xlabel('Re A')
ax.set_ylabel('Im A')
ax.legend()

ax = axes[1, 1]
ax.plot(W_bad.real, W_bad.imag, color='orange', linewidth=2)
ax.scatter([W_bad[0].real], [W_bad[0].imag], color='green', s=70, label='start')
ax.scatter([W_bad[-1].real], [W_bad[-1].imag], color='red', marker='x', s=90, label='end')
ax.set_aspect('equal', adjustable='box')
ax.set_title(f'Continued root, rho~{rho_bad:.3g}')
ax.set_xlabel('Re root')
ax.set_ylabel('Im root')
ax.legend()

fig.tight_layout()
plt.show()

print('Good zero-free disk case:')
print('  rho =', rho_good, 'trivial=', abs(rho_good - 1) < 1e-8)
print('Bad winding case:')
print('  rho =', rho_bad, 'trivial=', abs(rho_bad - 1) < 1e-8)
````

## Wider spectrum of special cases

The zero-free disk case is the easiest positive case. More generally, what matters is not the shape of the image region alone, but the winding of the loop $A_N(t)$ around zero.

The cases below illustrate the spectrum:

1. **Disk avoiding zero:** winding is zero. A single logarithm branch exists. Root monodromy is trivial.
2. **Small loop inside a zero-free annulus sector:** winding is still zero. A local logarithm branch exists on a simply connected sector. Root monodromy is trivial locally.
3. **Full annulus loop around zero:** winding is nonzero. Root monodromy is nontrivial unless the winding is divisible by the root degree.
4. **Multiple winding loop:** a cover of degree $q=d/\gcd(d,k)$ trivializes monodromy upstairs, but descent to the base happens only if the original monodromy was already trivial.

This is the story for the Böttcher basin: local zero-free charts are harmless, but loops whose $A_N$ image winds around zero create the actual obstruction.

````python
def loop_disk(center=2+0j, radius=0.4, samples=500):
    t = np.linspace(0, 2*np.pi, samples)
    return center + radius*np.exp(1j*t)

def loop_annulus_sector(radius=1.5, angle0=0.2, angle_width=0.8, samples=500):
    t = np.linspace(0, 2*np.pi, samples)
    # small closed loop inside a sector, not around zero
    return radius*np.exp(1j*(angle0 + angle_width/2 + 0.15*np.cos(t))) * (1 + 0.08*np.sin(t))

def loop_winding(radius=1.0, winding=1, samples=500):
    t = np.linspace(0, 2*np.pi, samples)
    return radius*np.exp(1j*winding*t)

cases = [
    ('disk avoiding 0', loop_disk(), 8),
    ('sector loop, no winding', loop_annulus_sector(), 8),
    ('annulus loop, winding 1', loop_winding(winding=1), 8),
    ('annulus loop, winding 2', loop_winding(winding=2), 8),
    ('annulus loop, winding 8', loop_winding(winding=8), 8),
]

fig, axes = plt.subplots(len(cases), 3, figsize=(14, 4*len(cases)))
summary = []
for row, (name, A_path, degree) in enumerate(cases):
    root_path = continued_root(A_path, degree)
    rho = root_path[-1] / root_path[0]
    wind = winding_number(A_path)
    k = round(wind)
    q = degree // math.gcd(degree, abs(k)) if k != 0 else 1
    descends = (k % degree == 0)
    summary.append((name, degree, wind, rho, q, descends))

    ax = axes[row, 0]
    ax.plot(A_path.real, A_path.imag, color='tab:red')
    ax.scatter([0], [0], color='black', s=25)
    ax.set_aspect('equal', adjustable='box')
    ax.set_title(f'{name}: A-loop')
    ax.set_xlabel('Re A')
    ax.set_ylabel('Im A')

    ax = axes[row, 1]
    ax.plot(root_path.real, root_path.imag, color='limegreen')
    ax.scatter([root_path[0].real], [root_path[0].imag], color='green', s=55, label='start')
    ax.scatter([root_path[-1].real], [root_path[-1].imag], color='red', marker='x', s=70, label='end')
    ax.set_aspect('equal', adjustable='box')
    ax.set_title(f'continued {degree}-th root')
    ax.set_xlabel('Re root')
    ax.set_ylabel('Im root')
    ax.legend(fontsize=8)

    ax = axes[row, 2]
    ax.axis('off')
    ax.text(0.02, 0.80, f'winding ≈ {wind:.2f}', fontsize=11)
    ax.text(0.02, 0.62, f'monodromy rho ≈ {rho:.3g}', fontsize=11)
    ax.text(0.02, 0.44, f'minimal cover q = {q}', fontsize=11)
    ax.text(0.02, 0.26, f'descends to base? {descends}', fontsize=11,
            color=('green' if descends else 'red'))
    if abs(rho - 1) < 1e-8:
        verdict = 'GOOD: trivial monodromy'
        color = 'green'
    else:
        verdict = 'OBSTRUCTION: branch changes'
        color = 'red'
    ax.text(0.02, 0.08, verdict, fontsize=12, color=color, weight='bold')

fig.tight_layout()
plt.show()

print('Summary table')
print('case | degree | winding | monodromy | cover q | descends')
for name, degree, wind, rho, q, descends in summary:
    print(f'{name:24s} | {degree:2d} | {wind:7.2f} | {rho:.4g} | {q:3d} | {descends}')
````

## General proof template suggested by the examples

The special cases suggest a local-to-global proof strategy.

1. Cover the image of each basin path under $A_N(z)=\Phi_\infty(f^N(z))$ by zero-free simply connected charts.
2. On each chart choose a logarithm branch and hence a $2^N$-root branch.
3. On overlaps compare the two branches by a locally constant element of $\mu_{2^N}$.
4. The product of these overlap multipliers around a loop is the monodromy element.
5. Prove this product is $1$ for the normalized Böttcher branch.

The examples show what each possible outcome means:

- zero winding: the proof is local and monodromy is trivial;
- nonzero winding: a cover can trivialize upstairs, but descent requires the monodromy element to be $1$ downstairs;
- full theorem: prove the basin's normalized pullback branch always lands in the descent/trivial case.

## Proof template with examples

The intended proof has a local-to-global shape. The finite root algebra is already formalized; the missing part is the analytic continuation argument.

### Template

**Step 1. Push the basin loop to the $A$-plane.**

For a basin loop $\gamma$, set

$$
A_N(t)=\Phi_\infty(f^N(\gamma(t))).
$$

**Step 2. Cover the image by zero-free simply connected charts.**

Choose open sets $U_i\subset\mathbb C\setminus\{0\}$ such that each $U_i$ has a single logarithm branch and

$$
A_N([t_i,t_{i+1}])\subset U_i.
$$

**Step 3. Choose local roots.**

On each chart define

$$
W_i(t)=\exp\left(\frac{1}{2^N}\mathrm{Log}_i(A_N(t))\right).
$$

**Step 4. Compare roots on overlaps.**

On an overlap, two local roots differ by a locally constant element

$$
\zeta_i\in\mu_{2^N}.
$$

**Step 5. Multiply overlap factors around the loop.**

The monodromy is

$$
\rho_N(\gamma)=\prod_i\zeta_i.
$$

**Step 6. Prove triviality.**

The branch descends to a single-valued basin coordinate exactly when

$$
\rho_N(\gamma)=1.
$$

The examples below show the template in four regimes.

````python
# Four examples for the proof template.
examples = [
    {
        'name': 'Single zero-free disk',
        'description': 'One chart covers the loop; no overlap monodromy.',
        'winding': 0,
        'degree': 8,
        'charts': 1,
    },
    {
        'name': 'Two overlapping zero-free charts',
        'description': 'Several charts, but total overlap multiplier is 1.',
        'winding': 0,
        'degree': 8,
        'charts': 2,
    },
    {
        'name': 'Loop winds once around zero',
        'description': 'Local charts exist, but overlap product is nontrivial.',
        'winding': 1,
        'degree': 8,
        'charts': 6,
    },
    {
        'name': 'Loop winds by multiple of degree',
        'description': 'Nonzero winding but monodromy is trivial modulo degree.',
        'winding': 8,
        'degree': 8,
        'charts': 6,
    },
]

fig, axes = plt.subplots(len(examples), 2, figsize=(12, 3.2*len(examples)))
for row, ex in enumerate(examples):
    degree = ex['degree']
    winding = ex['winding']
    t = np.linspace(0, 2*np.pi, 600)
    if winding == 0 and ex['charts'] == 1:
        A = 2 + 0.35*np.exp(1j*t)
    elif winding == 0:
        A = 1.5*np.exp(1j*(0.4 + 0.45*np.sin(t))) * (1 + 0.10*np.cos(2*t))
    else:
        A = np.exp(1j*winding*t)
    root = continued_root(A, degree)
    rho = root[-1] / root[0]
    trivial = abs(rho - 1) < 1e-8

    ax = axes[row, 0]
    ax.plot(A.real, A.imag, color='tab:red', linewidth=2)
    ax.scatter([0], [0], color='black', s=25)
    ax.set_aspect('equal', adjustable='box')
    ax.set_title(ex['name'])
    ax.set_xlabel('Re A_N')
    ax.set_ylabel('Im A_N')

    # Draw chart boundaries as colored arcs/points for the story.
    cut_indices = np.linspace(0, len(A)-1, ex['charts']+1, dtype=int)
    for j in range(ex['charts']):
        segment = A[cut_indices[j]:cut_indices[j+1]+1]
        ax.plot(segment.real, segment.imag, linewidth=4, alpha=0.25)

    ax = axes[row, 1]
    ax.plot(root.real, root.imag, color='limegreen', linewidth=2)
    ax.scatter([root[0].real], [root[0].imag], color='green', s=60, label='start')
    ax.scatter([root[-1].real], [root[-1].imag], color='red', marker='x', s=80, label='end')
    ax.set_aspect('equal', adjustable='box')
    ax.set_title(f'rho~{rho:.3g}; trivial={trivial}')
    ax.set_xlabel('Re root')
    ax.set_ylabel('Im root')
    ax.legend(fontsize=8)

fig.tight_layout()
plt.show()

print('Template summary')
for ex in examples:
    degree = ex['degree']
    winding = ex['winding']
    rho = cmath.exp(2j * math.pi * winding / degree)
    print(f"{ex['name']}: charts={ex['charts']}, winding={winding}, degree={degree}, rho={rho:.4g}, trivial={abs(rho-1)<1e-8}")
    print('  ', ex['description'])
````

## What Lean must prove

For the actual basin and the actual function

$$
A_N(z)=\Phi_\infty(f^N(z)),
$$

Lean must prove the following theorem, not merely inspect numerical loops:

1. every local root branch can be analytically continued along every `BasinLoop`;
2. the endpoint branch differs from the starting branch by a unique element of $\mu_{2^N}$;
3. these elements are compatible as $N$ changes;
4. for the normalized branch, all monodromy elements are $1$.
